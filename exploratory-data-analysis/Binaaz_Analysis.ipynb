# 3.Outlier detection (IQR or z-score) and processing, selection is justified

```
num_cols = df.select_dtypes(include="number").columns
```

```
df[num_cols].describe()
# lat,lng koordinatlardi orda outlier sayila bilecek deyerler yoxdu. Amma Azerbaycandan kenar koordinatlar daxil olmus ola biler.
```

```
outliers_cols=['price','unit_price','views','Otaq sayı','Sahə (m²)','Torpaq sahəsi']
```

```
# Outliers cox boyukdur ve normal paylanma deyil. Rigt skewed du ona ona gore iqr istifade etdim. Cunki iqr median ile hesablayir ve medianda outliers dan az tesirlenir mean e baxanda.
q1 = df[outliers_cols].quantile(0.25)
q3 = df[outliers_cols].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
```

```
lower_bound, upper_bound
```

```
outlier_count = ((df[outliers_cols] < lower_bound) |(df[outliers_cols] > upper_bound)).sum()

outlier_count
```

```
## Yuxardaki neticeye baxanda outliers gorunur bunlar hamsi sehv data deyil, meselen price unit_price cox baha villarin qiymetidir.
 # Silmemisden once baxmaq lazimdir
```

```
df[outliers_cols].describe()
```

```
df[df['Sahə (m²)']>500000].shape
```

```
df[df['price']>3000000].shape
```

```
df[df['Otaq sayı']>17].shape
```

```
df[df['Torpaq sahəsi']>3500].shape
```

```
df[outliers_cols].nlargest(20,'Sahə (m²)')
```

```
df[outliers_cols].nlargest(20,'price')
```

```
from sklearn.ensemble import IsolationForest  # coxsutunlu qeyri-normal datada IQR den daha stabil netice verir

# Torpaq sahesi çox NaN olduğu ucun ayri saxladim, esas sutunlarla isledim.
cols_for_outliers = ['price', 'unit_price', 'views', 'Otaq sayı', 'Sahə (m²)']

iso_f = IsolationForest(contamination=0.01, random_state=42)  #datanin ~1% outlier sayir.
iso_f.fit(df.loc[:, cols_for_outliers])
outlier_pred = iso_f.predict(df.loc[:, cols_for_outliers])

df.shape
```

```
outlier_mask = (outlier_pred == -1)
df.loc[outlier_mask, :]   # atilacaq setirler
```

```
df.loc[outlier_mask, :].shape
```

```
df = df.loc[outlier_pred == 1].reset_index(drop=True)
df.shape
```

# 5.Visualization (distribution graphs, correlation heatmap, categorical breakdown)

```
num_cols = df.select_dtypes(include='number').columns
num_cols
```

```
cols_to_plot = list(num_cols)
n = len(cols_to_plot)
n_cols = 2
n_rows = int(np.ceil(n / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 6, n_rows * 4))
axes = axes.flatten()

for ax, col in zip(axes, cols_to_plot):
    data = df[col].dropna()
    data = data[data > 0]  # log ucun musbet deyer lazimdi
    bins = np.logspace(np.log10(data.min()), np.log10(data.max()), 30)
    ax.hist(data, bins=bins)
    ax.set_xscale('log')
    ax.set_title(col)

for ax in axes[len(cols_to_plot):]:
    ax.axis('off')

plt.tight_layout()
plt.savefig(
    "numeric_cols_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```
corr = df.select_dtypes(include=["number"]).corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.savefig(
    "correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```
cat_cols = df.select_dtypes(include='category').columns
```

```
for col in cat_cols:
  plt.figure(figsize=(7,1))
  sns.countplot(
    data=df,
    y=col,
    order=df[col].value_counts().index)
  plt.show()

```

```
cat_cols = df.select_dtypes(include=['object', 'category']).columns

for col in cat_cols:
    print(f"{col}: {df[col].nunique()} unique")
```

```
cols_to_plot = [col for col in cat_cols if df[col].nunique() <= 20]

n = len(cols_to_plot)
n_cols = 2
n_rows = int(np.ceil(n / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 6, n_rows * 4))
axes = axes.flatten()

for ax, col in zip(axes, cols_to_plot):
    order = df[col].value_counts().index
    sns.countplot(data=df, x=col, order=order, color='lightgreen', ax=ax)
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)

for ax in axes[len(cols_to_plot):]:
    ax.axis('off')

plt.tight_layout()
plt.savefig(
    "object&categoric_cols_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
